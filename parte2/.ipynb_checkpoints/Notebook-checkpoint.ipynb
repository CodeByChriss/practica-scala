# Ejercicio 4 — Función para determinar si una mano se pasa de 21

Solución propuesta: Se siguen los pasos indicandos en el enunciado y se crea una función llamada bust que recibe la puntuación de una mano como Int y duelve un Boolean dependiendo de si es mayor que 21 o no.

```scala
def bust(hand: Int): Boolean = {
    if (hand > 21)
        true
    else
        false
}

println(bust(18))
println(bust(21))
println(bust(22))
println(bust(30))
```

# Ejercicio 5 — Comparación de dos manos

Solución propuesta: Se define una función maxHand en la que se reciben dos parámetros enteros (Int) y se comprueba cual es el mayor de los dos para ser devuelto. En caso de ser igual se devuelve el segundo parámetro ya que no importa si es el segundo o el primero porque son iguales.

```scala
def maxHand(a:Int, b:Int): Int = {
    if (a > b)
        a
    else
        b
}

println(maxHand(17,19))
println(maxHand(20,18))
println(maxHand(21,21))
```

# Explicación

## ¿Qué sucede cuando ambas manos tienen el mismo valor?
Por como he hecho el código unicamente comprueba si A es mayor que B y en este caso devuelve A y si no, devuelve B. Esto supone que si B es mayor que A (no cumple la condición del primer if) devuelve B y si son igual (tampoco cumple con la condición) devuelve B. En este último caso (cuando son iguales) da igual devolver A o B ya que tienen el mismo valor.

# Ejercicio 6 — Decidir el ganador de una partida

Solución propuesta: Se crea una función llamada ganador con dos parámetros enteros (Int) que devuelve un entero (Int). El entero que devuelve se decide con una serie de condiciones donde se acaba devolviendo 0 si ambos parámetros son superiores a 21, el primer parámetro si el segundo es mayor que 21 y se devuelve el segundo parámetro si el primero es mayor que 21 y en caso de que ninguno sea mayor de 21, se devuelve el parámetro más alto.

```scala
def ganador(handA: Int, handB: Int): Int = {
    if (handA > 21 && handB > 21){ // Condición 1
        0
    }else if (handA > 21){ // Condición 2
        handB
    }else if (handB > 21){ // Condición 3
        handA
    }else{ // Else 1
        if (handA > handB) // Condición 4
            handA
        else // Else 2
            handB
    }
}

println(ganador(26,20))
println(ganador(18,22))
println(ganador(24,25))
println(ganador(17,19))
println(ganador(21,20))
```

# Explicación

## 26 y 20
Cumple la condición 2

## 18 y 22
Cumple la condición 3

## 24 y 25
Cumple la condición 1

## 17 y 19
Cumple el Else 1 y el Else 2

## 21 y 20
Cumple el Else 1 y la condición 4

# Ejercicio 9 — Recorrer un Array con while

Solución propuesta: Se crea el array indicando su contenido y se crea la variable i que se usará como index en el while para recorrer el array y evitar un bucle infinito.

```scala
val manos = Array(17, 24, 21, 19, 26)
var i = 0

while (i < manos.length){
    val mano = manos(i)
    println(mano)
    println(bust(mano))
    i+=1
}
```

# Explicación

## ¿Por qué el contador debe ser mutable?
Para poder ir aumentando en 1 cada vez que pasa por el while.

## ¿Qué ocurriría si no incrementaras `i`?
Que el bucle while sería infinito y la aplicación no terminaría  nunca.

## ¿Qué elementos del ejercicio corresponden al estilo imperativo?
La variable mutable i (var i = 0). El bucle explícito (while (i < manos.length)) con todo su contenido.

# Ejercicio 10 — Listas e inmutabilidad

Solución propuesta: Se crea la lista principal (jugadores), se muestra y luego se crea una nueva agregando en la cabeza (como primer valor) un nuevo jugadores. Se muestran ambas listas y luego se muestra la longitud de la nueva lista y la misma lista pero en reversa.

```scala
val jugadores = List("Alex", "Chen", "Marta")
println(jugadores)

val nuevos_jugadores = "Sindhu" :: jugadores
println(jugadores)
println(nuevos_jugadores)

println(nuevos_jugadores.length)
println(nuevos_jugadores.reverse)
```

# Explicación

## ¿Por qué la operación con `::` produce una lista nueva en lugar de modificar la lista original?
Las listas en Scala no se mueve modificar ni en tamaño ni sus elementos por lo que se debe generar una nueva lista sí o sí.

# Ejercicio 12 — Operadores relacionales y lógicos

Solución propuesta: Se ha hecho un println para comprobar el resultado False o True de cada expresión.

```scala
val handA = 18
val handB = 21
val handC = 25

println(handA > handB)
println(handB == 21)
println(handC != 21)
println(handA <= 21 && handB <= 21)
println(handA > 21 || handC > 21)
println(handB > 21)
```


| Expresión | Resultado | Explicación |
| --- | --- | --- |
| handA > handB | false | 18 no es mayor que 21 |
| handB == 21 | true | handB que es 21 es igual que 21 |
| handC != 21 | true | handC que es 25 no es igual que 21 |
| handA <= 21 && handB <= 21 | true | 18 y 21 es menor o igual que 21 |
| handA > 21 \|\| handC > 21 | true | 18 o 25 es mayor que 21 |
| handB > 21 | false | 21 no es mayor que 21 |

# Ejercicio 13 — foreach y funciones como valores

Solución propuesta: Se ha creado un array usando los datos indicados en el enunciado y se ha creado una variable i para recorrer el array en un bucle while y luego se ha usado el método foreach para recorrer el array.

```scala
val manos = Array(17, 24, 21, 26, 18)

var i = 1

while (i < manos.length){
    println(bust(manos(i)))
    i+=1
}

manos.foreach(bust) // Esto no muestra nada ya que la función devuelve el valor
manos.foreach(mano => println(bust(mano)))
```

# Explicación

## ¿Cuál necesita un contador?
La opción A, el while es el que necesita un contador.

## ¿Cuál necesita una variable `var` para recorrer la colección?
La opción A, el while es el que necesita una variable que sirva de contador.

## ¿Cuál se aproxima más al estilo funcional presentado en el material del curso (Datacamp)?
La opción B, el foreach.

# Ejercicio 14 — Efectos secundarios y estilo de programación

Solución propuesta: Se crea una variable total que contendrá el valor que se le sume que haya sido pasado a la función sumarAlTotal. Se llama varias veces a la función sumarAlTotal y se muestra para ver como cambia el contenido de la variable total. Se crea otra función sumar que devuelve la suma de a+b siendo a y b parámetros de la propia función.

```scala
var total = 0

def sumarAlTotal(valor: Int) = {
  total = total + valor
}

sumarAlTotal(2)
println(total)

sumarAlTotal(6)
println(total)

sumarAlTotal(3)
println(total)

def sumar(a: Int, b: Int): Int = {
    a+b
}

println(sumar(2,3))
```

| Característica | sumarAlTotal | sumar |
| --- | --- | --- |
| Modifica datos externos | Sí, modifica la variable global total | No, trabaja solo con argumentos locales |
| Devuelve un resultado calculado | No, devuelve Unit / () | Sí, devuelve un valor de tipo Int |
| Utiliza efecto secundario | Sí, mutación del estado de la memoria | No, no tiene efecto secundarios |
| Estilo predominante | Imperativo | Funcional / Declarativo |

# Ejercicio 15 — Programa integrado: torneo de Twenty-One

Solución propuesta: Se definen los Arrays y Listas con los datos indicados. Se crea la función comprobarManos utilizando un bucle while e instrucciones if-else jutno a la función anteriormente creada bust para verificar e imprimir si cada jugador se pasa o no de 21. También, se crea la función manoMasAlta que filtra las puntuaciones válidas (<=21) y determina la puntuación más alta. Finalmente se repite con manosRonda2.

```scala
val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manos = Array(18, 24, 21, 20)

// Comprobamos si cada mano se pasa o no de 21
def comprobarManos(manos: Array[Int], jugadores: List[String]) ={
    var i = 0
    while (i < manos.length){
        val jugador = jugadores(i)
        val mano = manos(i)
    
        val estado = bust(mano)
        
        if (!estado){
            println(s"La mano de $jugador es de $mano y no se pasa de 21.")
        }else{
            println(s"La mano de $jugador es de $mano y se pasa de 21.")
        }
        
        i+=1
    }
}

comprobarManos(manos, jugadores)

// Determinamos cuál es la puntiación más alta
// Esta función devuelve el index de la mano más alta
def manoMasAlta(manos: Array[Int], jugadores: List[String]) = {
    var i = 0
    var mano_mas_alta = -1
    var index_mano_mas_alta = -1
    var jugador_mano_mas_alta = ""
    while (i < manos.length){
        val mano = manos(i)
        
        if ((mano_mas_alta == -1 || mano_mas_alta < mano) && mano <= 21){
            mano_mas_alta = mano
            index_mano_mas_alta = i
            jugador_mano_mas_alta = jugadores(i)
        }
        
        i+=1
    }

    println(s"La mano más alta es de $jugador_mano_mas_alta siendo de $mano_mas_alta.")
}

manoMasAlta(manos, jugadores)

// Ronda 2
println("----------------RONDA 2-----------------")
val manosRonda2 = Array(22, 19, 20, 21)
comprobarManos(manosRonda2, jugadores)
manoMasAlta(manosRonda2, jugadores)
```

# Explicación

## ¿Qué partes de tu solución son mutables?
Las variables de control dentro de las funciones declaradas con var: la variable i (contador del bucle while) y las variables de seguimiento mano_mas_alta, index_mano_mas_alta y jugador_mano_mas_alta. También son mutables los datos de los Arrays de manos y manosRonda2.

## ¿Qué partes son inmutables?
La lista jugadores declarada como List[String], cuyo tamaño y contenido no se pueden modificar y las referencias declaradas con val: jugadores, manos, manosRonda2 y las constantes locales dentro de las iteraciones (jugador, mano, estado).

## ¿Dónde utilizas funciones?
Funciones propias creadas: comprobarManos y manoMasAlta.
Función de otros ejercicios: bust(mano).

## ¿Dónde utilizas estructuras de control?
Bucles: Estructura while utilizada para iterar sobre los índices de los arrays. Condicionales: Estructuras if y else para evaluar los estados booleanos y verificar si la puntuación es válida e igual o superior a la máxima registrada.

## ¿Qué parte de tu solución consideras más cercana al estilo funcional?
El uso de la lista inmutable jugadores sin modificar la colección original y la llamada a la función bust(mano), que toma un parámetro y devuelve una evaluación sin alterar el estado externo.

## ¿Qué parte consideras más cercana al estilo imperativo?
El uso explícito de bucles while con variables contadoras mutables (var i = 0, i += 1). Las variables mutables internas (var mano_mas_alta) asignadas paso a paso mediante efectos secundarios dentro del bucle. Y el uso de las funciones comprobarManos y manoMasAlta para imprimir directamente por consola (println) retornando Unit.
